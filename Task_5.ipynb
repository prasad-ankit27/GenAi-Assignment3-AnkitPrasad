```python
prices=[100,250,400,1200,50,2000,850]
def great(value):
    if value>500:
        return True

def less(value):
    if value<=500:
        return True
    
print('Prices greater than 500:',list(filter(great,prices)))
print('Prices less than or equal to 500:',list(filter(less,prices)))

```
